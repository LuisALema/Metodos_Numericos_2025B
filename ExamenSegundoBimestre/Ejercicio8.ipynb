```python
import numpy as np
import matplotlib.pyplot as plt
from src import gauss_jacobi, gauss_seidel
```

```python
def trayectoria_gs(A, b, x0, tol=1e-6, it=10):
    tray = [x0.reshape(-1,1)]
    for k in range(it):
        x_new = gauss_seidel(A=A, b=b, x0=tray[-1], tol=tol, max_iter=2)
        tray.append(x_new)
    return np.hstack(tray).T
```

```python
def dibujar(tray, titulo):
    x = tray[:,0]
    y = tray[:,1]

    plt.plot(x, y, "o:")
    plt.plot(x[0], y[0], "gd", label="X0")
    plt.plot(x[-1], y[-1], "ro", label="X*")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title(titulo)
    plt.legend()
    plt.grid()
    plt.show()
```

# Ejercicio 1

```python
A = np.array([[2,3],[1,2]],float)
b = np.array([5,8],float)
x0 = np.array([0,0],float)

tray1 = trayectoria_gs(A,b,x0)
dibujar(tray1,"Ejercicio 1")

sol1 = tray1[-1]
m1 = (tray1[-1,1]-tray1[0,1])/(tray1[-1,0]-tray1[0,0])

print("Solución:", sol1)
print("Pendiente m =", m1)
```
