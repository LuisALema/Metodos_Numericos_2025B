## Introducción

Los formatters en Python permiten mantener un código legible, ordenado y acorde a estándares como PEP8. 
Estas herramientas automatizan el formateo del código sin alterar su funcionamiento, facilitando el trabajo colaborativo y el mantenimiento del software.


## Objetivo

Analizar y comparar diferentes formatters populares en Python aplicados sobre un mismo archivo de código, 
con el fin de identificar sus características y diferencias principales.


## Herramientas Utilizadas

- Python 3  
- Jupyter Notebook  
- Black  
- autopep8  
- YAPF  
- isort

## Formatter Black

Black aplica un formato estricto y automático, imponiendo un estilo uniforme sin necesidad de configuración manual.


```python
!black formatter_test.py

with open("formatter_test.py", "r", encoding="utf-8") as f:
    print(f.read())
```

## Formatter autopep8

autopep8 corrige errores de estilo de acuerdo con la norma PEP8, manteniendo la estructura original del código.


```python
!autopep8 --in-place formatter_test.py

with open("formatter_test.py", "r", encoding="utf-8") as f:
    print(f.read())
```

## Formatter YAPF

YAPF permite estilos configurables y genera un código con mayor flexibilidad visual.


```python
!yapf -i formatter_test.py

with open("formatter_test.py", "r", encoding="utf-8") as f:
    print(f.read())
```

## Formatter isort

isort ordena y organiza automáticamente las sentencias de importación del archivo Python.


```python
!isort formatter_test.py

with open("formatter_test.py", "r", encoding="utf-8") as f:
    print(f.read())
```
