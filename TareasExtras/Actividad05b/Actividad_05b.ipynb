## Introducción

El Prompt Engineering es la habilidad de diseñar instrucciones claras para interactuar con modelos de lenguaje. 
Un mal prompt produce respuestas pobres, mientras que un prompt bien diseñado permite obtener resultados útiles, coherentes y precisos.
Este informe se basa en el video *"You suck at prompting"* y en documentación oficial de OpenAI, Google y Anthropic.


## Objetivo

Identificar y explicar las recomendaciones más importantes para mejorar la calidad de los prompts al trabajar con modelos de lenguaje artificial.


## ¿Qué es un prompt?

Un prompt no es solo una pregunta, sino un **programa en lenguaje natural** que le indica a un modelo qué hacer, cómo hacerlo y en qué formato entregar el resultado.
Los modelos de lenguaje no razonan como humanos, sino que predicen la continuación más probable del texto.


## 1. Pensar el prompt como un programa

El usuario debe estructurar su solicitud como una serie de instrucciones claras.  
Mientras más preciso sea el prompt, menos probabilidades tendrá la IA de generar respuestas incorrectas.


## 2. Definir una persona o rol

Asignar una identidad a la IA (por ejemplo: ingeniero, profesor, analista) enfoca el tipo de conocimiento que utilizará para responder.
Esto reduce resultados genéricos.


## 3. Proporcionar todo el contexto

El modelo no sabe lo que el usuario no le dice.  
La falta de contexto provoca que la IA “rellene” información, lo que genera errores o datos inventados.


## 4. Definir el formato de salida

Es importante indicar si se quiere una lista, un texto corto, una tabla, un tono formal o informal, o un límite de palabras.
Esto estandariza y mejora la calidad del resultado.


## 5. La meta-habilidad: claridad de pensamiento

El principal problema no es la IA, sino la falta de claridad del usuario.
Si no se puede explicar bien una idea, la IA tampoco podrá ejecutarla correctamente.


## Conclusión

El Prompt Engineering no se trata de trucos, sino de aprender a expresar ideas con precisión.
Entre más claro sea el pensamiento del usuario, mejores serán las respuestas de la inteligencia artificial.


## Recomendación Final

Se recomienda practicar la escritura de prompts estructurados y crear una biblioteca personal de instrucciones bien diseñadas.
