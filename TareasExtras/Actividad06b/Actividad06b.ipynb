## Introducción

Los modelos Transformers utilizan grandes matrices de pesos para procesar secuencias.
Estas matrices pueden ser muy costosas en memoria y tiempo de cómputo.
El factoreo de matrices permite descomponer estas matrices en factores más pequeños,
reduciendo costos sin afectar significativamente la precisión del modelo.


## Objetivo

Explicar el uso del factoreo de matrices dentro de la arquitectura Transformer,
indicando sus principales razones y ventajas en modelos de redes neuronales profundas.


## ¿Qué es el factoreo de matrices?

El factoreo de matrices consiste en descomponer una matriz grande en el producto
de dos o más matrices de menor dimensión, conservando la información principal.
Ejemplos comunes son SVD, factorización de bajo rango y descomposición matricial.


## Uso de matrices en Transformers

En los Transformers, las matrices aparecen en:
- Proyecciones Q, K y V en la atención.
- Capas Feed Forward.
- Matrices de embeddings.
Estas matrices suelen tener millones de parámetros.


## Razón 1: Reducción de parámetros

El factoreo permite representar una matriz grande como el producto de dos matrices pequeñas,
reduciendo drásticamente el número de pesos del modelo.


## Razón 2: Aceleración del entrenamiento

Menos parámetros implican menos multiplicaciones matriciales,
lo que disminuye el tiempo de entrenamiento y el consumo de energía.


## Razón 3: Menor uso de memoria

El modelo ocupa menos memoria en GPU o RAM, permitiendo entrenar modelos grandes
en hardware más limitado.


## Ventaja 1: Eficiencia computacional

Las operaciones se vuelven más rápidas y ligeras,
mejorando el rendimiento general del sistema.


## Ventaja 2: Regularización implícita

El bajo rango actúa como una forma de regularización,
reduciendo el sobreajuste del modelo.


## Ventaja 3: Modelos más escalables

Permite construir Transformers más grandes sin que el costo computacional
crezca de forma exponencial.


## Conclusión

El factoreo de matrices es una técnica clave para hacer los Transformers
más eficientes, rápidos y escalables, sin sacrificar notablemente su precisión.


## Recomendación Final

Se recomienda aplicar factorización de bajo rango en arquitecturas grandes
para optimizar recursos y mejorar la viabilidad del modelo en producción.
