## Introducción

Las arquitecturas Hopper y Blackwell de NVIDIA representan dos generaciones
de GPUs diseñadas para acelerar cargas de trabajo de inteligencia artificial,
cómputo científico y aprendizaje profundo. 
Estas arquitecturas priorizan el rendimiento, la eficiencia energética
y el soporte para múltiples representaciones numéricas.


## Objetivo

Analizar las diferencias principales entre Hopper y Blackwell, así como
el impacto de los formatos numéricos en el rendimiento de modelos de IA.


## Arquitectura Hopper

Hopper (H100) introdujo Tensor Cores de cuarta generación,
soporte para TP32 y optimización para entrenamiento de modelos grandes.
Está orientada a centros de datos y supercómputo.


## Arquitectura Blackwell

Blackwell (B100/B200) es la evolución de Hopper.
Incorpora Tensor Cores más eficientes, mayor ancho de banda
y soporte extendido para formatos de baja precisión.


## Diferencia entre FP32 y TF32

FP32 (32 bits) usa una mantisa completa, dando máxima precisión.  
TF32 (19 bits efectivos) reduce la precisión para acelerar operaciones
en IA, manteniendo un error aceptable.


## Representaciones numéricas soportadas

Ambas arquitecturas soportan:
- FP64: alta precisión científica  
- FP32: precisión estándar  
- TF32: entrenamiento acelerado  
- FP16 / BF16: entrenamiento rápido  
- INT8 / INT4: inferencia eficiente  


## Preferencia por menor precisión

Los modelos de IA toleran errores pequeños.
Reducir precisión permite:
- Más operaciones por segundo
- Menor consumo de energía
- Menor uso de memoria


## Ventajas clave de Blackwell sobre Hopper

- Mayor rendimiento por watt  
- Mejor soporte para INT y FP16  
- Más eficiencia en modelos gigantes  


## Conclusión

Blackwell mejora a Hopper en velocidad, eficiencia y manejo de formatos
numéricos de baja precisión, haciendo posible entrenar modelos de IA
mucho más grandes y rápidos.
