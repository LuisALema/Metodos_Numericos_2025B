## Introducción

El módulo logging de Python permite registrar mensajes durante
la ejecución de un programa. Se usa para detectar errores,
monitorear procesos y guardar información relevante.


## Objetivo

Conocer y utilizar el módulo logging en Python para registrar
mensajes en consola, archivos y servicios externos como Telegram.


## ¿Qué es logging?

Es una librería estándar de Python que permite generar mensajes
con distintos niveles de severidad.


## Celda de código: logging en consola


```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(filename)s | %(message)s"
)

logging.info("Mensaje en consola")
logging.error("Mensaje de error")
```

## Celda de código: logging en archivo


```python
file_handler = logging.FileHandler("registro.log")

formatter = logging.Formatter(
    "%(asctime)s | %(levelname)s | %(filename)s | %(message)s"
)

file_handler.setFormatter(formatter)

logger = logging.getLogger()
logger.addHandler(file_handler)

logger.warning("Mensaje guardado en archivo")
```

## Celda de código: logging en Telegram


```python
import requests

TOKEN = "TU_TOKEN"
CHAT_ID = "TU_CHAT_ID"

def enviar_telegram(mensaje):
    url = f"https://api.telegram.org/bot{TOKEN}/sendMessage"
    data = {"chat_id": CHAT_ID, "text": mensaje}
    requests.post(url, data=data)

enviar_telegram("Mensaje enviado desde logging")
```

```python
from colorama import Fore, Style, init
init()

logging.info(Fore.GREEN + "Mensaje verde" + Style.RESET_ALL)
logging.error(Fore.RED + "Mensaje rojo" + Style.RESET_ALL)
```
